```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Clustal Omega identity matrix data
matrix_data = {
    'Human': [100.00, 85.40, 86.20],
    'Mouse': [85.40, 100.00, 94.80],
    'Rat': [86.20, 94.80, 100.00]
}
row_labels = ['Human (NP_002244.1)', 'Mouse (NP_034742.2)', 'Rat (NP_037194.2)']
column_labels = ['Human', 'Mouse', 'Rat']

df = pd.DataFrame(matrix_data, index=row_labels, columns=column_labels)

plt.figure(figsize=(7, 5), dpi=100)
sns.heatmap(df, annot=True, fmt=".2f", cmap="YlGnBu", linewidths=1.5, 
            cbar_kws={'label': 'Sequence Identity (%)'}, annot_kws={"size": 11, "weight": "bold"})

plt.title('VEGFR-2 Pairwise Sequence Identity (Clustal Omega)', fontsize=12, fontweight='bold', pad=12)
plt.tight_layout()

# Image automatically aapke folder mein save ho jayegi
plt.savefig('vegfr2_msa_heatmap.png', dpi=300)
plt.show()
```
