```python
import numpy as np 
import pandas as pd 
import Bio as biopython 

from Bio import Entrez, SeqIO

Entrez.email = "shreyaupadhyay1702@gmail.com"  # Fixed variable name
accession_id = input("enter the accession id ")
custom_name = input("enter the custom name")
file_name = f"{custom_name}.fasta"  # Removed \n from filename

try:
    print(f"\nNCBI_ID : {accession_id}, extracting")
    handle = Entrez.efetch(db="protein", id=accession_id, rettype="fasta", retmode="text")
    record = SeqIO.read(handle, "fasta")
    handle.close()
    with open(file_name, "w") as file:
        SeqIO.write(record, file, "fasta")
    print("\nfile saved successfully")
    print(f"protein_name: {record.description}")
    print(f"total length: {len(record.seq)} amino_acid")
    print(f"file saved as: {file_name}")
    print("\n----------- whole sequence ------------")
    print(record.seq)
except Exception as e:
    print("\nError: Data not fetched")
    print("reason:", e)
```
